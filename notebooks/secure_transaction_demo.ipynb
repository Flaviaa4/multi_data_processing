# Secure transaction demo

This notebook runs the same face-authorized transaction flow as the CLI. An unauthorized result stops before transaction details or product recommendations are shown.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / 'src').is_dir():
    ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT / 'src'))

from demo_app import run_transaction, DEMO_PROFILES
```

```python
authorized_image = ROOT / 'data' / 'images' / 'flavia.jpeg'
authorized = run_transaction(authorized_image)
authorized
```

```python
from PIL import Image
from tempfile import NamedTemporaryFile

with NamedTemporaryFile(suffix='.png', delete=False) as file:
    unauthorized_path = Path(file.name)
Image.new('RGB', (64, 64), color=(12, 24, 48)).save(unauthorized_path)
try:
    unauthorized = run_transaction(unauthorized_path, threshold=1.01)
finally:
    unauthorized_path.unlink(missing_ok=True)
unauthorized
```
